```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(data["Item"])

y = data["Waste_Type"]

model = LogisticRegression(C=20, max_iter=1000)

model.fit(X, y)

X.shape
```

```python
def classify(text):
    new_X = vectorizer.transform([text])

    if new_X.nnz == 0:
        return "Not sure - no known words. Ask a staff member."

    category = model.predict(new_X)[0]
    row = waste_types[waste_types["Waste_Type"] == category].iloc[0]
    confidence = model.predict_proba(new_X).max()
    return category, row["Bin_Color"], row["Hazardous"], confidence
```
